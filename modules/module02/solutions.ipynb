## 1. Overlapping chunks

Chunk 13 words in windows of 5 with overlap 2; stop when the final window reaches the end. Explain why words are not tokenizer tokens.

```python
def chunk(words,width,overlap):
    if not 0<=overlap<width: raise ValueError("Invalid overlap")
    out=[];start=0
    while start<len(words):
        end=min(start+width,len(words));out.append((start,end,words[start:end]))
        if end==len(words):break
        start=end-overlap
    return out
chunks=chunk(list(range(13)),5,2)
assert [(a,b) for a,b,_ in chunks]==[(0,5),(3,8),(6,11),(9,13)]
assert chunk([],5,2)==[]
print(chunks)
```

## 2. All-source recall

For a two-source question, compare finding one required source with finding both. Compute both coverage and the capstone all-source success.

```python
required={"spec","policy"}
def score(found):return {"coverage":len(required&set(found))/len(required),"all_source_success":int(required<=set(found))}
assert score(["spec"])=={"coverage":.5,"all_source_success":0}
assert score(["spec","policy"])["all_source_success"]==1
print(score(["spec"]),score(["spec","policy"]))
```

## 3. Citation is not support

Detect unknown citation IDs. Show a known ID attached to an unsupported assertion and explain why syntax cannot establish faithfulness.

```python
import re
context={"spec":"Warranty: 24 months."}
answer="Warranty is lifetime [spec]. The item was made yesterday [unknown]."
cited=set(re.findall(r"\[([^]]+)\]",answer));unknown=cited-set(context)
assert unknown=={"unknown"}
review={"unknown_ids":sorted(unknown),"known_citation_supported":False,"reason":"24 months does not support lifetime."}
print(review)
```

## 4. Refusal review

A response contains the refusal phrase followed by an invented serial number. Preserve a human review field instead of automatically calling it a successful refusal.

```python
answer="I don't have enough information. Your serial number is ZX-123."
record={"phrase_match":"don't have enough information" in answer.lower(),"meaningful_refusal":None,"raw":answer}
assert record["phrase_match"] and record["meaningful_refusal"] is None
record.update(meaningful_refusal=False,reason="Invented continuation contradicts the refusal.")
print(record)
```
