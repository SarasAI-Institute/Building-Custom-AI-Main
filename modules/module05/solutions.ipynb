## 1. GOOD short-circuit

Use deterministic fake components to prove a GOOD verdict bypasses retrieval and analysis. This is a contract test, not real project integration.

```python
calls=[]
def pipeline(verdict):
    note={"verdict":verdict,"description":"fixture"}
    result={"note":note,"evidence":"","report":""}
    if verdict=="GOOD":return result
    calls.append("retrieve");result["evidence"]="[fixture] observed crack"
    calls.append("analyst");result["report"]="Fixture report"
    return result
result=pipeline("GOOD");assert calls==[] and result["report"]==""
result=pipeline("DEFECTIVE");assert calls==["retrieve","analyst"] and isinstance(result["evidence"],str)
print(result)
```

## 2. Source-to-product mapping

Filter applicable sources using external product identity. A high similarity score must not move an unrelated product into the answer.

```python
sources={"kettle_spec":["KT-2000"],"toaster_spec":["TS-400"],"returns":["KT-2000","TS-400"]}
hits=[{"source":"toaster_spec","score":.99},{"source":"kettle_spec","score":.8},{"source":"returns","score":.7}]
selected=[h for h in hits if "KT-2000" in sources.get(h["source"],[])]
assert [h["source"] for h in selected]==["kettle_spec","returns"]
print(selected)
```

## 3. Failures stay in the batch

Summarize four attempted cases: correct GOOD, supported report, parser failure and runtime failure. Count case success only from an explicit reviewed decision.

```python
records=[{"status":"good","success":True},{"status":"report","success":True},{"status":"parse_error","success":False},{"status":"runtime_error","success":False}]
assert all(type(r["success"]) is bool for r in records)
rate=sum(r["success"] for r in records)/len(records)
assert rate==.5
print({"attempts":len(records),"reviewed_success_rate":rate,"failures":[r["status"] for r in records if not r["success"]]})
```

## 4. Artifact identity

Show that a different quantization arm or changed checkpoint invalidates a saved integration decision, even if the path name is unchanged.

```python
import hashlib,json
def sha(x):return hashlib.sha256(json.dumps(x,sort_keys=True).encode()).hexdigest()
plan={"arm":"folded_int8","files":{"model.safetensors":"example-sha"},"split":"frozen-v1"}
review={"plan_sha256":sha(plan),"chosen_arm":"folded_int8"}
assert review["plan_sha256"]==sha(plan)
changed={**plan,"arm":"folded_nf4"};assert review["plan_sha256"]!=sha(changed)
print("Changed arm requires its own evaluated decision.")
```
